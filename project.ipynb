```python
from pulp import *

# Create the model
model = LpProblem("Profit_Maximization", LpMaximize)

# Decision Variables
x = LpVariable('Product_A', lowBound=0)
y = LpVariable('Product_B', lowBound=0)

# Objective Function (Maximize Profit)
model += 20*x + 30*y

# Constraints
model += 2*x + 4*y <= 100  # Raw material constraint
model += 3*x + 2*y <= 90   # Labor constraint

# Solve the model
model.solve()

# Output results
print("Status:", LpStatus[model.status])
print("Product A =", value(x))
print("Product B =", value(y))
print("Maximum Profit =", value(model.objective))
```

Problem Statement : A company produces two products A and B. The goal is to maximize profit.
Objective : Maximize profit = 20x + 30y
Constraints:
2x + 4y ≤ 100 (raw material)
3x + 2y ≤ 90 (labor)
Decision Variables:
x = Number of Product A
y = Number of Product B

Product B gives more profit
Resources are limited
Best combination gives maximum profit

```python
import matplotlib.pyplot as plt

# Values from output
products = ['Product A', 'Product B']
values = [value(x), value(y)]

# Create bar chart
plt.bar(products, values)

# Labels and title
plt.xlabel("Products")
plt.ylabel("Quantity")
plt.title("Optimal Production Plan")

# Show graph
plt.show()
```

Graph Analysis:

The bar chart shows the optimal production quantities of Product A and Product B.
It helps to visually understand which product is produced more.
